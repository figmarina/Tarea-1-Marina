## Estadísticos descriptivos

```r
# Longitud total (CT)

mean(datos_mugil$CT, na.rm = TRUE)

median(datos_mugil$CT, na.rm = TRUE)

sd(datos_mugil$CT, na.rm = TRUE)

min(datos_mugil$CT, na.rm = TRUE)

max(datos_mugil$CT, na.rm = TRUE)

# Coeficiente de variación (%)
sd(datos_mugil$CT, na.rm = TRUE) /
  mean(datos_mugil$CT, na.rm = TRUE) * 100
```

## Función estadística

```r
# Función de estadística descriptiva

estadistica_basica <- function(x) {  x <- x[!is.na(x)]
  c(
    N = length(x),
    Media = mean(x),
    Mediana = median(x),
    SD = sd(x),
    Minimo = min(x),
    Maximo = max(x),
    CV = sd(x) / mean(x) * 100  )}

# Aplicar a la longitud total
estadistica_basica(datos_mugil$CT)

# Aplicar al peso total
estadistica_basica(datos_mugil$PT)
```

## Función para detectar outliers

```r
# Función para detectar outliers mediante IQR

detectar_outliers <- function(x) {

  Q1 <- quantile(x, 0.25, na.rm = TRUE)
  Q3 <- quantile(x, 0.75, na.rm = TRUE)

  rango <- IQR(x, na.rm = TRUE)

  LI <- Q1 - 1.5 * rango
  LS <- Q3 + 1.5 * rango

  which(x < LI | x > LS)}

# Aplicar a la longitud total
detectar_outliers(datos_mugil$CT)

# Aplicar al peso total
detectar_outliers(datos_mugil$PT)
```
