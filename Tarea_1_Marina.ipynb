## Descripción del conjunto de datos

El conjunto de datos contiene información biométrica y reproductiva de peces identificados como *Mugil curema*, organizada en una hoja de cálculo Excel.

La base presenta **721 registros y 13 variables**, entre ellas medidas corporales, pesos, índices biológicos, sexo, estadio de maduración y fechas de colecta y procesamiento.

Los datos permiten realizar análisis exploratorios de características biológicas y reproductivas.
